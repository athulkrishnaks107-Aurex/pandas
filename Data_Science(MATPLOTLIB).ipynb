```python
import matplotlib.pyplot as plt

x=[3,2,2.5,3,3.5,4,3]
y=[0.5,2,3,2.25,3,2,0.5]

plt.plot(x,y)
plt.fill_between(x,y,color="red")
plt.title("HEART")
plt.show()
```

```python
x=[3,3.5,4,2.75,4.25,3]
y=[1,2.75,1,2,2,1]

plt.plot(x,y)
plt.fill_between(x,y,color="orange")
plt.show()
```

```python
months=["jan","feb","mar","apr","may"]
sales=[20,30,25,40,50]

plt.plot(months,sales)
plt.show()
```

```python
months=["jan","feb","mar","apr","may"]
sales=[20,30,25,40,50]
profit=[5,8,6,12,15]

plt.plot(months,sales,label="sales")      #A piece of text that identifies or describes a single specific axis.
plt.plot(months,profit,label="profit")
plt.legend()                           # it displays the label of the line.
plt.show()
```

```python
from matplotlib.lines import lineStyles
months=["jan","feb","mar","apr","may"]
sales=[20,30,25,40,50]
plt.plot(months,sales)

plt.xlabel("months")
plt.ylabel("sales")
plt.title("monthly sales")                  #used to display the title
plt.grid(True)                              #used to display the grid
plt.plot(months,sales,marker="o",color="red",linestyle="--")

plt.show()
```

```python
from matplotlib.lines import lineStyles
months=["jan","feb","mar","apr","may"]
sales=[20,30,25,40,50]
plt.bar(months,sales)

plt.xlabel("months")                                                      #add .bar instead of plot to show in bar style
plt.ylabel("sales")
plt.title("monthly sales")
plt.grid(True)
plt.bar(months,sales,color="red",linestyle="--")


plt.show()
```

```python
study_hours=[1,2,3,4,5,6]
marks=[40,45,55,65,75,85]

plt.scatter(study_hours,marks)                         #it shows points only without line
plt.title("study hours vs marks")
plt.xlabel("study hours")
plt.ylabel("marks")

plt.show()
```

```python
marks=[45,50,52,55,60,62,65,68,70,72,75,78,80,82,85,90]

plt.hist(marks)

plt.title("marks distribution")
plt.xlabel("marks")
plt.ylabel("frequency")

plt.show()
```

```python
categories=["food","travel","shopping",'others']
amount=[4000,2000,3000,1000]

plt.pie(amount,labels=categories)
plt.title("Monthly Expenses")

plt.show()
```

```python
fig,ax=plt.subplots(2,1)             #FIG MEANS TO SHOW 2 OR MORE FIGURES AND SUBPLOTS(ROW ,COLUMN)

ax[0].plot([1,2,3],[10,20,30])            #ax[0] MEANS FIRST FIGURE
ax[1].bar(["A","B","C"],[20,15,30])       #ax[1] means second figure

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=[2,4,6,8,10]

sns.barplot(x=x,y=y)
plt.show()
```

```python
data={
    "product":["laptop","phone","tablet","laptop","phone","tablet"],
    "sales":[50,80,40,70,90,60],
    "price":[50000,30000,20000,55000,32000,22000],
    "month":["jan","jan","feb","feb","mar","mar"]
}

sns.barplot(x="product",y="sales",data=data)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data=np.random.randn(1000)
plt.figure(figsize=(8,6))
sns.histplot(data,bins=30,kde=True,color="purple",edgecolor="black")

plt.xlabel("values")
plt.ylabel("frequency")
plt.title("Distribution Of Random Data Using Seaborn")
plt.grid(axis="y",alpha=0.75)

plt.show()
```

```python
products=['laptop',"phone","tablet"]
sales=[50,80,40]

plt.pie(sales,labels=products,autopct="%1.2f%%")
#autopct is a parameter used to automatically calculate and display percentage values to a pie chart
plt.title("sales distribution")

plt.show()
```
