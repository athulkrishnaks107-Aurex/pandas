```python
import pandas as pd
data=[20,30,40,50]
s=pd.Series(data)      #creating data with series
print(s)
```

```python
import pandas as pd
data=[20,30,40,50]
s=pd.Series(data,index=['a','b','c','d'])                 #adding index values as a,b,c,d
print(s)
```

```python
data={"name":"athul","age":19}
s=pd.Series(data)
print(s)
```

```python
data=[1,2,3,4,5]
s=pd.Series(data)
print(s.loc[1])                      #loc is used to call a specific position
```

```python
data=[1,2,3,4,5]
s=pd.Series(data)
print(s.loc[1:4:1])                      #it can be use as slicing
```

```python
data=[1,2,3,4,5]
s=pd.Series(data,index=['a','b','c','d','e'])
print(s.iloc[1:2])
```
