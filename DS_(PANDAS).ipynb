```python

import pandas as pd
data=[20,30,40,50]
s=pd.Series(data)      #creating data with series
print(s)
```

```python
import pandas as pd
data=[20,30,40,50]
s=pd.Series(data,index=['a','b','c','d'])                 #adding index values as a,b,c,d
print(s)
```

```python
data=[1,2,3,4,5]
s=pd.Series(data)
print(s.loc[1])                      #loc is used to call a specific position
```

```python
data=[1,2,3,4,5]
s=pd.Series(data)
print(s.loc[1:4:1])                      #it can be use as slicing
```

```python
data=[1,2,3,4,5]
s=pd.Series(data,index=['a','b','c','d','e'])
print(s.iloc[:2])
```

```python
data=[10,20,30,40,50]
s=pd.Series(data,index=['a','b','c','d','e'])
a=s[s>30]
print(a)
print(s.loc['a':'b'])
print(s.iloc[:2])
```

```python
sales=pd.Series([1200,850,1500,950,2000],index=["Monday","Tuesday","Wednesday","Thursday","Friday"])
print(sales)

max=sales.max()
print("\n Maximum sales =",max)

min=sales.min()
print("\n Minimum sales =",min)

lessthan1000=sales[sales<1000]
print("\n Days with sales less than 1000 =",lessthan1000)
```

```python
import pandas as pd
data=[20,20,30,30,50,60,60,80]
ar=pd.Series(data)
print(ar)
print("\n",ar.unique())                 #it shows unique numbers once only
print("\n",ar.nunique())                #it shows total unique numbers
print("\n",ar.value_counts())           #it shows how many times a number is repeated
```

```python
data=[70,None,30,None,60,60,80]
ar=pd.Series(data)
print(ar)
print('\n',ar.isnull())        #it shows the given value is null or not
print('\n',ar.notnull())       #it shows the given value is number or not
print('\n',ar.fillna(0))       #it fill none values by given number (0)
print('\n',ar.dropna())        #it delet none values and then print
```
