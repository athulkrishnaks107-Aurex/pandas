```python
import pandas as pd
data=[20,20,30,30,50,60,60,80]
ar=pd.Series(data)
print(ar)
print("\n",ar.unique())                 #it shows unique numbers once only
print("\n",ar.nunique())                #it shows total unique numbers
print("\n",ar.value_counts())           #it shows how many times a number is repeated
```

```python
data=[70,None,30,None,60,60,80]
ar=pd.Series(data)
print(ar)
print('\n',ar.isnull())        #it shows the given value is null or not
print('\n',ar.notnull())       #it shows the given value is number or not
print('\n',ar.fillna(0))       #it fill none values by given number (0)
print('\n',ar.dropna())        #it delet none values and then print
```

```python
import numpy as np
data={"name":["alice","bob","charlie","david","emma"],
      "age":[20,21,19,22,20],
      "marks":[85,75,np.nan,90,65],
      "attendance":[95,80,75,np.nan,88],
      "city":['kochi',"delhi","mumbai","kochi","chennai"]
}

df=pd.DataFrame(data)
print(df)


#BASIC INFORMATION

print("\nFirst 3 rows:")
print(df.head(3))

print("\nLast 2 rows: ")
print(df.tail(2))

print("\nshape= ",df.shape)

print("\ncolumns= ",df.columns)

print("\ndatatype= ",df.dtypes)

print("\nName and marks:")
print(df[["name","marks"]])

print("\nRows 1 to 3,Name and Marks(using loc):")           #loc can print a specific position
print(df.loc[0:3,["name","marks"]])

print("\nRows 1 to 3(usingiloc)")                           #iloc cannot print a specific position,it printsn the whole values
print(df.iloc[0:3])

print("\nstudents with marks greater than 75:")
print(df[df["marks"]>75])

print("\nstudents from kochi:")
print(df[df["city"]=="kochi"])

print("\nstudents with no attendance:")
print(df[df["attendance"].isnull()])

print("\nstudents with marks > 70 AND attendance > 80:")
print(df[(df["marks"]>70) & (df["attendance"]>80)])


#-----MATHEMATICAL OPERATIONS-----

print("\nmarks + 5:")
print(df["marks"]+5)

print("\nmarks * 2:")
print(df["marks"]*2)

print("\nAverage marks:")
print(df["marks"].mean())

print("\nMaximum marks:")
print(df["marks"].max())

print("\nMinimum marks:")
print(df["marks"].min())

print("\ntotal marks:")
print(df["marks"].sum())

#-----------RENAMING-----------

df_rename = df.rename(columns={"marks":"scores",
                               "attendance":"attendance_percentage"})
print("\renamed dataframe:")
print(df_rename)

print("\nmissing values:")
print(df.isna())

print("\nnumber of missing values:")
print(df.isna().sum())

#-------FORWARD FILL-------
df_forward=df.copy()
df_forward=df_forward.ffill()
print("\nforward fill:")
print(df_forward)

#-------BACKWARD FILL-------
df_backward=df.copy()
df_backward=df_backward.fillna(method="bfill")
print("\nbackward fill:")
print(df_backward)
```

'FORWARD FILL' IS USED TO FILL THE NULL VALUE WITH THE PREVIOS VALUE

'BACKWARD FILL' IS USED TO FILL THE NULL VALUES WITH VALUE AFTER NULL
